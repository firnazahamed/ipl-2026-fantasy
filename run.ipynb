### Build Historical Ownership GSheet

```python
# ── Build historical ownership GSheet (2019–2025) ─────────────────────────────
# Pre-requisite: create a new Google Sheet, share it with the service account,
# then paste its URL as HIST_OWN_URL below and in settings.py.
# Run this cell once; re-run whenever you want to refresh historical data.
# Note: throttled to ~43 reads/min to stay under the 60/min quota — takes ~3 min.

import time
import gspread
from settings import hist_squads_by_year, owner_team_dict, service_account_credentials

HIST_YEARS   = [2019, 2020, 2021, 2022, 2023, 2024, 2025]
HIST_OWN_URL = "https://docs.google.com/spreadsheets/d/1_euk8aJLKVaRp4OFe0FDaydgxMCu-gpZnSfTE0Hr6no/"

_SLEEP = 1.4  # seconds between API reads — keeps us at ~43/min, safely under 60/min

_known_owners = set(owner_team_dict.keys())
_STATS_HEADERS = {
    "name", "player", "player name", "player_name", "name_batting",
    "name_bowling", "team", "role", "category", "cat", "runs", "wickets",
    "matches", "points", "total", "total_points", "batting_points",
    "bowling_points", "fielding_points", "price", "nationality", "s.no",
    "sno", "sr", "economy", "econ", "balls", "overs", "catches",
    "stumpings", "owner", "unsold",
}

def _is_owner_col(col_s):
    if col_s in _known_owners:
        return True
    if "_" in col_s:
        return False
    try:
        int(col_s)
        return False
    except ValueError:
        pass
    return col_s.lower() not in _STATS_HEADERS

# Use gspread directly (no Streamlit cache layer in notebook context)
gc = gspread.service_account(filename=service_account_credentials)

def _read_tab(spreadsheet_url, tab_name):
    """Read a GSheet tab directly via gspread with retry on 429."""
    for attempt in range(4):
        try:
            sh = gc.open_by_url(spreadsheet_url)
            ws = sh.worksheet(tab_name)
            data = ws.get_all_values()
            time.sleep(_SLEEP)
            if not data:
                return None
            import pandas as pd
            return pd.DataFrame(data[1:], columns=data[0])
        except gspread.exceptions.APIError as e:
            if e.response.status_code == 429:
                wait = 60 * (attempt + 1)
                print(f"  Rate limited — waiting {wait}s before retry {attempt+1}/3...")
                time.sleep(wait)
            else:
                raise
    return None

def _list_tabs(spreadsheet_url):
    """List worksheet names with retry on 429."""
    for attempt in range(4):
        try:
            sh = gc.open_by_url(spreadsheet_url)
            tabs = [ws.title for ws in sh.worksheets()]
            time.sleep(_SLEEP)
            return tabs
        except gspread.exceptions.APIError as e:
            if e.response.status_code == 429:
                wait = 60 * (attempt + 1)
                print(f"  Rate limited — waiting {wait}s before retry...")
                time.sleep(wait)
            else:
                raise
    return []

# player (original casing) → {year: set(owners)}
ownership = {}

for year in HIST_YEARS:
    url = hist_squads_by_year.get(year)
    if not url:
        print(f"  No URL for {year}, skipping")
        continue

    all_tabs = _list_tabs(url)
    # Only process week tabs (ignore stats/other tabs)
    tabs = [t for t in all_tabs if t.lower().startswith("week")]
    print(f"Year {year}: {len(tabs)}/{len(all_tabs)} week tabs — reading...", end=" ", flush=True)

    for tab in tabs:
        df = _read_tab(url, tab)
        if df is None or df.empty:
            continue
        for col in df.columns:
            col_s = str(col).strip()
            if not col_s or not _is_owner_col(col_s):
                continue
            for val in df[col].astype(str):
                player = val.strip()
                if not player or player.lower() in ("nan", "", "0"):
                    continue
                if player not in ownership:
                    ownership[player] = {}
                if year not in ownership[player]:
                    ownership[player][year] = set()
                ownership[player][year].add(col_s)

    print("done")

# Build DataFrame
import pandas as pd
rows = []
for player in sorted(ownership.keys()):
    row = {"Player": player}
    for year in HIST_YEARS:
        owners_set = ownership[player].get(year, set())
        row[str(year)] = ", ".join(sorted(owners_set)) if owners_set else ""
    rows.append(row)

hist_own_df = pd.DataFrame(rows)[["Player"] + [str(y) for y in HIST_YEARS]]
print(f"\nBuilt {len(hist_own_df)} player rows")
display(hist_own_df.head(10))

# Write to GSheet
sh  = gc.open_by_url(HIST_OWN_URL)
ws  = sh.get_worksheet(0)
ws.clear()
data = [hist_own_df.columns.tolist()] + hist_own_df.values.tolist()
ws.update("A1", data)
print(f"Written to: {HIST_OWN_URL}")
```
